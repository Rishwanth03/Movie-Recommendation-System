# Movie Recommendation EDA
Explore the MovieLens metadata and inspect the structure before building the recommender.

```python
import pandas as pd
from pathlib import Path

root = Path('..')
movies = pd.read_csv(root / 'data' / 'movies.csv')
print('movies shape:', movies.shape)
print(movies.head().to_string(index=False))
print('nulls:\n', movies.isna().sum())
print('duplicate rows:', movies.duplicated().sum())

if (root / 'data' / 'ratings.csv').exists():
    ratings = pd.read_csv(root / 'data' / 'ratings.csv')
    print('ratings shape:', ratings.shape)
    print(ratings.head().to_string(index=False))
else:
    print('ratings.csv not found; continuing with movie metadata available in this project.')
```

```python
movies['year'] = movies['title'].str.extract(r'\((\d{4})\)$')
movies['year'] = pd.to_numeric(movies['year'], errors='coerce')

print('valid year values:', movies['year'].notna().sum())
print('earliest year:', int(movies['year'].min()))
print('latest year:', int(movies['year'].max()))

genres = movies['genres'].fillna('').str.split('|').explode()
genres = genres[genres != '']
print('top genres:')
print(genres.value_counts().head(10))
```

```python
import matplotlib.pyplot as plt

ratings_path = root / 'data' / 'ratings.csv'
if ratings_path.exists():
    ratings = pd.read_csv(ratings_path)
    print('rating summary:')
    print(ratings['rating'].describe())
    plt.figure(figsize=(8, 5))
    ratings['rating'].plot(kind='hist', bins=20, edgecolor='black')
    plt.title('Rating Distribution')
    plt.xlabel('Rating')
    plt.ylabel('Count')
    plt.tight_layout()
    plt.show()
else:
    print('Skipped rating distribution because ratings.csv is not available yet.')
```
